```python
states = ['S0', 'S1', 'S3', 'S4', 'G']
actions = ['Up', 'Down', 'Left', 'Right']
gamma = 0.9
positions = {
    'S0': (0, 0), 'S1': (0, 1), 'G': (0, 2),
    'X': (1, 0), 'S3': (1, 1), 'S4': (1, 2)
}
def transition(state, action):
    if state == 'G':
        return 'G'
    row, col = positions[state]
    if action == 'Up':
        row -= 1
    elif action == 'Down':
        row += 1
    elif action == 'Left':
        col -= 1
    elif action == 'Right':
        col += 1
    if row < 0 or row > 1 or col < 0 or col > 2:
        return state
    for s, pos in positions.items():
        if pos == (row, col):
            return state if s == 'X' else s
    return state
def reward(state, action, next_state):
    if next_state == 'G':
        return 10
    return -1
policy = {
    'S0': 'Right',
    'S1': 'Down',
    'S3': 'Right',
    'S4': 'Up'
}
def state_value_function(state):
    total = 0
    discount = 1
    while state != 'G':
        action = policy[state]
        next_state = transition(state, action)
        r = reward(state, action, next_state)
        total += discount * r
        discount *= gamma
        state = next_state
    return total
def action_value_function(state, action):
    next_state = transition(state, action)
    r = reward(state, action, next_state)
    if next_state == 'G':
        return r
    return r + gamma * state_value_function(next_state)
print("State Value V(S0):", state_value_function('S0'))
print("Action Value Q(S0, Up):",
      action_value_function('S0', 'Up'))
```

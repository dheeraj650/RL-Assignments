```python
import random

machines = ['A', 'B', 'C']

N_play = 10

epsilon = 0.2

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0, 'B': 0, 'C': 0}

total_reward = 0


for play in range(1, N_play + 1):

    r = random.random()


    max_Q = max(Q.values())

    best_machines = [m for m in machines if Q[m] == max_Q]

    if r < epsilon:

        exploration_machines = [
            m for m in machines if m not in best_machines
        ]

        selected_machine = random.choice(exploration_machines)

        decision = "Exploration"

    else:

        selected_machine = random.choice(best_machines)

        decision = "Exploitation"

    print("\nPlay", play)
    print("Random number:", round(r, 4))
    print("Decision:", decision)
    print("Play with machine", selected_machine)

    reward = float(input("Enter reward: "))


    N[selected_machine] += 1

    R[selected_machine] += reward


    Q[selected_machine] = (
        Q[selected_machine]
        + (1 / N[selected_machine])
        * (reward - Q[selected_machine])
    )

    total_reward += reward

    print("Updated Q-value of", selected_machine, "=",
          round(Q[selected_machine], 2))


print("\n" + "=" * 50)
print("FINAL RESULTS")
print("=" * 50)

print("\nMachine\tN\tR\tQ")

for machine in machines:
    print(
        machine,
        "\t",
        N[machine],
        "\t",
        R[machine],
        "\t",
        round(Q[machine], 2)
    )

print("\nTotal Reward =", total_reward)
```

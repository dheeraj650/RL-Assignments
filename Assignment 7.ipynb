```python
# Assigment - 1
n = int(input("Enter the number of time steps: "))
rewards = []
for i in range(n):
    reward = float(input(f"Enter reward for time step {i + 1}: "))
    rewards.append(reward)
gamma = float(input("Enter the discount factor (gamma): "))
undiscounted_return = sum(rewards)
discounted_return = 0
for i in range(n):
    discounted_return += (gamma ** i) * rewards[i]
print("\nEntered reward values:", rewards)
print("Undiscounted Return:", undiscounted_return)
print("Discounted Return:", discounted_return)
```

```python
#Assignment - 2
%%writefile grid_env_det.py
states = ['S0', 'S1', 'S3', 'S4', 'G']
actions = ['Up', 'Down', 'Left', 'Right']
gamma = 0.9

positions = {
    'S0': (0, 0),
    'S1': (0, 1),
    'G':  (0, 2),
    'X':  (1, 0),
    'S3': (1, 1),
    'S4': (1, 2)
}

def transition(state, action):
    if state == 'G':
        return 'G'

    row, col = positions[state]

    if action == 'Up':
        row -= 1
    elif action == 'Down':
        row += 1
    elif action == 'Left':
        col -= 1
    elif action == 'Right':
        col += 1

    if row < 0 or row > 1 or col < 0 or col > 2:
        return state

    for s, pos in positions.items():
        if pos == (row, col):
            if s == 'X':
                return state
            return s

    return state

def reward(state, action, next_state):
    if next_state == 'G':
        return 10
    return -1
```

```python
import grid_env_det as env

current_state = 'S1'
action = 'Right'

next_state = env.transition(current_state, action)
r = env.reward(current_state, action, next_state)

print("Current State:", current_state)
print("Action:", action)
print("Next State:", next_state)
print("Reward:", r)
print("Discount Factor:", env.gamma)
```

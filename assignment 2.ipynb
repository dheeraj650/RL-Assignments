```python
import random
machines = ['A', 'B', 'C']

N_play = 10

epsilon = 0.2

N = {
    'A': 0,
    'B': 0,
    'C': 0
}

R = {
    'A': [],
    'B': [],
    'C': []
}

Q = {
    'A': 0.0,
    'B': 0.0,
    'C': 0.0
}

total_reward = 0
for play in range(1, N_play + 1):

    r = random.random()

    if r < epsilon:
        # Exploration
        machine = random.choice(machines)
        decision = "Exploration"
    else:
        # Exploitation
        max_q = max(Q.values())

        best_machines = [
            m for m in machines
            if Q[m] == max_q
        ]

        machine = random.choice(best_machines)
        decision = "Exploitation"

    print(f"\nPlay {play}")
    print(f"Random number = {r:.2f}")
    print(f"Decision = {decision}")
    print(f"Selected Machine = {machine}")
    reward = float(input("Enter reward: "))

    N[machine] += 1
    R[machine].append(reward)

    Q[machine] = Q[machine] + (
        (reward - Q[machine]) / N[machine]
    )

    total_reward += reward
    print(f"N({machine}) = {N[machine]}")
    print(f"Q({machine}) = {Q[machine]:.2f}")
print("\n========== FINAL RESULTS ==========")

print("\nN Table:")
for machine in machines:
    print(f"N({machine}) = {N[machine]}")

print("\nR Table:")
for machine in machines:
    print(f"R({machine}) = {R[machine]}")

print("\nQ Table:")
for machine in machines:
    print(f"Q({machine}) = {Q[machine]:.2f}")

print(f"\nTotal Reward after 10 plays = {total_reward:g}")
```

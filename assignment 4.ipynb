```python
import math

machines = ['A', 'B', 'C']

N_play = 10

N = {machine: 0 for machine in machines}
R = {machine: [] for machine in machines}
Q = {machine: 0.0 for machine in machines}
UCB = {machine: 0.0 for machine in machines}

total_reward = 0

print("Upper Confidence Bound Multi-Armed Bandit")
print("-" * 50)

for t in range(1, N_play + 1):

    print(f"\nPlay {t}")
    for machine in machines:
        if N[machine] == 0:
            UCB[machine] = float('inf')
        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(t)) / N[machine]
            )
    print("\nUCB Values:")
    for machine in machines:
        if UCB[machine] == float('inf'):
            print(f"{machine}: Infinity")
        else:
            print(f"{machine}: {UCB[machine]:.3f}")
    selected_machine = max(machines, key=lambda x: UCB[x])

    print(f"\nPlay with machine {selected_machine}")
    reward = float(input("Enter reward: "))
    N[selected_machine] += 1
    R[selected_machine].append(reward)
    Q[selected_machine] = Q[selected_machine] + (
        (reward - Q[selected_machine]) / N[selected_machine]
    )
    total_reward += reward

    print(f"Updated Q-value of {selected_machine}: "
          f"{Q[selected_machine]:.3f}")

print("\n" + "=" * 60)
print("FINAL RESULTS")
print("=" * 60)

print(f"{'Machine':<10}{'N':<10}{'Rewards':<25}{'UCB':<12}{'Q-value'}")
print("-" * 75)

for machine in machines:

    if UCB[machine] == float('inf'):
        ucb_value = "Infinity"
    else:
        ucb_value = f"{UCB[machine]:.3f}"

    print(
        f"{machine:<10}"
        f"{N[machine]:<10}"
        f"{str(R[machine]):<25}"
        f"{ucb_value:<12}"
        f"{Q[machine]:.3f}"
    )

print("-" * 75)
print(f"\nTotal Reward Received = {total_reward}")
```
